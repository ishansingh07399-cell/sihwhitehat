# AI Model Evaluation

This notebook demonstrates the evaluation of the AI Models used in the Criminal Network Intel platform.
It proves that the Named Entity Recognition (NER) and Intent Classification (Zero-Shot) models are performing genuine inference rather than using hardcoded rules.

```python
import sys
import os
sys.path.append(os.path.abspath('..'))

from models.nlp_extractor import ner_pipeline, zero_shot_classifier
```

### 1. Named Entity Recognition (NER)

```python
text = "Suspect Jonathan Vercetti transferred $50,000 from Vice City to a bank in Liberty City."
entities = ner_pipeline(text)
for ent in entities:
    print(f"{ent['entity_group']}: {ent['word']} (Confidence: {ent['score']:.2f})")
```

### 2. Intent Classification (Zero-Shot) to predict IPC Sections

```python
report_text = "We have intercepted communications indicating a massive financial scam and money laundering operation."
candidate_labels = ["murder or lethal violence", "financial fraud or money laundering", "criminal conspiracy or coordinated planning", "illegal weapons or firearms"]

result = zero_shot_classifier(report_text, candidate_labels)
print(f"Predicted Intent: {result['labels'][0]}")
print(f"Confidence: {result['scores'][0]:.2f}")
```
